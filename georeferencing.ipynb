```python
from osgeo import gdal
```

```python
def copy_georeferencing(pan_image_path, pansharpened_image_path, output_image_path):
    # Open the panchromatic image
    pan_ds = gdal.Open(pan_image_path, gdal.GA_ReadOnly)
    if not pan_ds:
        raise FileNotFoundError(f"Could not open the panchromatic image at {pan_image_path}")

    # Read geotransform and projection from the panchromatic image
    geotransform = pan_ds.GetGeoTransform()
    projection = pan_ds.GetProjection()

    # Open the pansharpened image
    pansharpened_ds = gdal.Open(pansharpened_image_path, gdal.GA_ReadOnly)
    if not pansharpened_ds:
        raise FileNotFoundError(f"Could not open the pansharpened image at {pansharpened_image_path}")

    # Create a copy of the pansharpened image to write the georeferenced information
    driver = gdal.GetDriverByName('GTiff')
    output_ds = driver.CreateCopy(output_image_path, pansharpened_ds, 0)

    # Set the geotransform and projection
    output_ds.SetGeoTransform(geotransform)
    output_ds.SetProjection(projection)

    # Close the datasets
    pan_ds = None
    pansharpened_ds = None
    output_ds = None

    print(f"Georeferencing information copied successfully to {output_image_path}")
```

```python
# Define the paths
pan_image_path = "D:/geoinformatics/MajorProject/main/highres_patches/panpatchesarcu16/WG_PAN_CLIP_BLOCK_0_1.tif"
pansharpened_image_path = "D:/geoinformatics/MajorProject/main/output_tif/bymethods/Wavelet/Wavelet_0_1.tiff"
output_image_path = "D:/geoinformatics/MajorProject/main/output_tif/bymethods/Wavelet/georef/georef_Wavelet_0_1.tif"

# Copy georeferencing information
copy_georeferencing(pan_image_path, pansharpened_image_path, output_image_path)
```
