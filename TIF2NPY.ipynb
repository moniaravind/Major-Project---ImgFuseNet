```python
import rasterio
import numpy as np

# Path to the input .tif file and output .npy file
input_file = "D:/geoinformatics/MajorProject/main/output_tif/1_3/PanNet_georeferenced_.tif"
output_file = "D:/geoinformatics/MajorProject/main/npyfiles/1_3_PanNet_georeferenced"

# Open the .tif fil3
with rasterio.open(input_file) as src:
    # Read the entire raster data
    data = src.read()

# Save the data to a .npy file
np.save(output_file, data)

print(f"Raster data from {input_file} has been saved to {output_file}")
```
